```python
import re

# Improved marker phrases with regex patterns
marker_patterns = [
    r'^The present (?:study|work|report|manuscript|paper|research|investigation|article)\b',
    r'^In this (?:study|work|report|paper|research|investigation|article)\b',
    r'^This (?:study|work|report|paper|research|investigation|article)\b',
    r'^In the present (?:study|work|report|manuscript|paper|research|investigation|article)\b',
    r'^In the current (?:study|work|report|manuscript|paper|research|investigation|article)\b',
    r'^The aim of this (?:study|work|report|manuscript|paper|research|investigation|article)\b',
    r'^The purpose of this (?:study|work|report|manuscript|paper|research|investigation|article)\b',
    r'^The goal of this (?:study|work|report|manuscript|paper|research|investigation|article)\b',
    r'^Our (?:study|work|report|manuscript|paper|research|investigation|article)\b(?:used|employed|showed)\b',
    r'^We (?:used|employed|showed|investigated|present|report)\b',
    r'^It was (?:thus|therefore) our goal\b',
    r'^This (?:study|work|report|manuscript|paper|research|investigation|article) (?:was|is)\b',
    r'^Therefore, in this (?:study|work|report|manuscript|paper|research|investigation|article)\b',
    r'^Thus, in this (?:study|work|report|manuscript|paper|research|investigation|article)\b',

    r'^Here we\b',
    r'^Here, we\b',
    r'^The research reported here\b',
    r'^Inspired by this work, we\b',
    r'^Based on the (?:above )?reasons, we\b',
    r'^Herein,\b',
    r'^Recently, we\b',
    # Expanded verb patterns
    r'^We (?:describe|discuss|examine|analyze|present|investigate|report|show)\b',

]
# Compile patterns for better performance
compiled_patterns = [re.compile(pattern, re.IGNORECASE) for pattern in marker_patterns]
```

```python

def get_first_clean_word(text):
    text = re.sub(r'^([IVXLCDM]+)\.', 
                 lambda m: str(sum(1 for _ in m.group(1))) + '.', 
                 text,
                 flags=re.IGNORECASE)
    clean_start = re.sub(r'^[^a-zA-Z]*', '', text.lower())
    first_word = clean_start.split()[0] if clean_start else ""
    return re.sub(r'[^a-zA-Z]', '', first_word)

def process_introduction_section(section_content, compiled_patterns, section_starters):
    """Processes a section to extract relevant introduction content."""
    # Check if section qualifies as introduction
    first_word = get_first_clean_word(section_content[0])
    valid_section = first_word in section_starters
    
    if len(section_content) > 1 and not valid_section:
        second_word = get_first_clean_word(section_content[1])
        valid_section = second_word in section_starters
    
    if not valid_section:
        return None
    
    # Process the content
    new_text = ''.join(section_content)
    cleaned_text = re.sub(r'\.\s+', '.', new_text)
    
    # Split sentences
    sentences = []
    start = 0
    for boundary in [m.start() for m in re.finditer(r'\.([A-Z])', cleaned_text)]:
        sentences.append(cleaned_text[start:boundary+1])
        start = boundary + 1
    sentences.append(cleaned_text[start:])
    
    # Find first marker sentence
    start_index = None
    for i, sentence in enumerate(sentences):
        if any(pattern.match(sentence.strip()) for pattern in compiled_patterns):
            start_index = i
            break
    
    # Get relevant sentences
    if start_index is not None:
        relevant_sentences = sentences[start_index:]
    else:
        relevant_sentences = sentences[-6:] if len(sentences) >= 6 else sentences
    
    return ' '.join(relevant_sentences)
```

```python
def total_size(obj, seen=None):
    if seen is None:
        seen = set()
    obj_id = id(obj)
    if obj_id in seen:
        return 0
    seen.add(obj_id)
    size = sys.getsizeof(obj)
    if isinstance(obj, (list, tuple, set, frozenset)):
        size += sum(total_size(i, seen) for i in obj)
    elif isinstance(obj, dict):
        size += sum(total_size(k, seen) + total_size(v, seen) for k, v in obj.items())
    return size

list_var = original_text
dict_var = filtered_text 

print(f"Total size of list: {total_size(list_var)} bytes")
print(f"Total size of dict: {total_size(dict_var)} bytes")
```

```python
import re

def replace_non_uppercase_newlines(text):
    """
    Replaces newlines with spaces when:
    - The newline is NOT preceded by an uppercase letter AND
    - The newline is NOT followed by an uppercase letter
    (i.e., only replaces when both sides are lowercase/non-letters)
    Preserves newlines near uppercase letters (sentence boundaries).
    """
    # Combined pattern using negative lookarounds
    return re.sub(
        r'(?<![A-Z])\n(?![A-Z])', 
        ' ', 
        text, 
        flags=re.MULTILINE
    )

# Example usage:
sample_text = """This is a test\nof the emergency\nBroadcast system.
Here's another\nexample with Mixed\nCase letters.
Numbers like 1\n2 should stay.
But lowercase\nconnections should merge.
UPPERCASE\nCONNECTION stays.\n"""
print(replace_non_uppercase_newlines(sample_text))
```

```python
import re

def remove_dna_sequences(text):
    """
    Removes DNA sequences in the format 5′-XXXXXXX-3′ from text.
    """
    # Pattern explanation:
    # 5′       - literal 5′
    # -?       - optional hyphen
    # [ACGT]+  - one or more DNA nucleotides (A, C, G, T)
    # -?       - optional hyphen
    # 3′       - literal 3′
    return re.sub(r'5′[-]?[ACGTNacgnt]+[-]?3′', '', text)

# Example usage:
sample_text = "This contains a DNA sequence 5′-GAGACACAATTCATCGATGGTTAGTG-3′ to remove."
clean_text = remove_dna_sequences(sample_text)
print(clean_text)  # Output: "This contains a DNA sequence  to remove."
```

```python
import os
import json
import time
import logging
import pandas as pd
import subprocess
import concurrent.futures
from io import BytesIO
from pdfminer.high_level import extract_text_to_fp
from pdfminer.layout import LAParams

# === Configuration ===
DOI_FILE = './temp_pdfs/dois.txt'
RESULTS_FILE = './temp_pdfs/result.csv'
PDF_DIR = './temp_pdfs'
BATCH_SAVE_INTERVAL = 5
BATCH_TIMEOUT = 1800  # seconds
PDF_TIMEOUT = 1200   # seconds (20 minutes)

# === Logging Setup ===
logging.basicConfig(
    level=logging.INFO,
    format='%(message)s',
)

# === State ===
full_texts = {}
failed_batches = []
failed_pdfs = []
batch_counter = 0
save_index = 0

# === Text Extraction Helper ===
def extract_text_safe(file_path):
    with open(file_path, 'rb') as data:
        output = BytesIO()
        laparams = LAParams(word_margin=0.2)
        extract_text_to_fp(data, output, laparams=laparams)
        return output.getvalue().decode()

# === Main Process ===
logging.info("Starting DOI processing pipeline...")

for i, batch in enumerate(batched_art_dois):
    batch_start = time.time()
    logging.info(f"Processing batch {i + 1} with {len(batch)} DOIs...")

    try:
        # Write DOIs to file
        with open(DOI_FILE, 'w') as f:
            for doi in batch:
                f.write(f"{doi}\n")
        logging.debug(f"Written {len(batch)} DOIs to {DOI_FILE}")

        # Wait 1 minute before PyPaperBot
        logging.info("Waiting 10 seconds before running PyPaperBot...")
        time.sleep(10)

        # Run PyPaperBot with timeout
        logging.info("Running PyPaperBot subprocess...")
        subprocess.run(
            f'python -m PyPaperBot --doi-file={DOI_FILE} --dwn-di="{PDF_DIR}" --restrict="1"',
            shell=True,
            timeout=BATCH_TIMEOUT
        )

        # Read result CSV
        try:
            results = pd.read_csv(RESULTS_FILE).dropna(subset=['PDF Name'])
            logging.info(f"Retrieved {len(results)} results with valid PDFs.")
        except Exception as e:
            logging.error(f"Failed to read {RESULTS_FILE}: {e}")
            failed_batches.append(batch)
            continue

        for found_doi, pdf in results.set_index('DOI')['PDF Name'].items():
            file_path = os.path.join(PDF_DIR, pdf)
            logging.info(f"Extracting text for DOI {found_doi} from {pdf}...")

            pdf_start = time.time()
            try:
                with concurrent.futures.ThreadPoolExecutor(max_workers=1) as executor:
                    future = executor.submit(extract_text_safe, file_path)
                    text = future.result(timeout=PDF_TIMEOUT)

                duration = time.time() - pdf_start
                logging.info(f"Text extraction for DOI {found_doi} completed in {duration:.2f} seconds")

                text = cut_unrelevant_info(text)
                full_texts[found_doi] = text

            except concurrent.futures.TimeoutError:
                logging.warning(f"Text extraction for DOI {found_doi} timed out after {PDF_TIMEOUT} seconds")
                failed_pdfs.append(found_doi)
            except Exception as e:
                logging.error(f"Failed to extract text for {file_path}: {e}")
                failed_pdfs.append(found_doi)

        # Clean up temporary PDFs
        logging.info("Cleaning up temporary PDF files...")
        subprocess.run(f'rm -f {PDF_DIR}/*.pdf', shell=True)

        batch_counter += 1

        # Save every N batches
        if batch_counter % BATCH_SAVE_INTERVAL == 0:
            out_file = f'full_texts_v2_part_{save_index}.json'
            with open(out_file, 'w') as f:
                json.dump(full_texts, f)
            logging.info(f"Saved batch data to {out_file}")
            full_texts = {}
            save_index += 1

        duration = time.time() - batch_start
        logging.info(f"Finished batch {i + 1} in {duration:.2f} seconds")

    except subprocess.TimeoutExpired:
        logging.error(f"Batch {i + 1} timed out after {BATCH_TIMEOUT} seconds")
        failed_batches.append(batch)
    except Exception as e:
        logging.error(f"Unexpected error during batch {i + 1}: {e}")
        failed_batches.append(batch)

# Save remaining full_texts
if full_texts:
    out_file = f'full_texts_part_{save_index}.json'
    with open(out_file, 'w') as f:
        json.dump(full_texts, f)
    logging.info(f"Saved remaining data to {out_file} (final)")

# Save failed batches
if failed_batches:
    with open('failed_batches.json', 'w') as f:
        json.dump(failed_batches, f)
    logging.warning(f"{len(failed_batches)} batches failed and were saved to failed_batches.json")

# Save failed PDFs
if failed_pdfs:
    with open('failed_pdfs.json', 'w') as f:
        json.dump(failed_pdfs, f)
    logging.warning(f"{len(failed_pdfs)} individual PDFs failed and were saved to failed_pdfs.json")

logging.info("DOI processing complete.")
```

```python


# -----------------------------------------------------------
# 2nd-pass: re-run the batches that failed the first time
# -----------------------------------------------------------
import json, os, time, subprocess, concurrent.futures, pandas as pd
from io import BytesIO
from pdfminer.high_level import extract_text_to_fp
from pdfminer.layout import LAParams

# ---- Load the failed batch list (skip if the file is missing) ----
if not os.path.exists('failed_batches.json'):
    print("Nothing to retry – 'failed_batches.json' not found.")
    exit()

with open('failed_batches.json', 'r') as fh:
    retry_batches = json.load(fh)

print(f"Retrying {len(retry_batches)} batches from failed_batches.json …")

# ---- Re-initialise state just for the retry pass ----
full_texts      = {}
failed_batches2 = []      # failures on the 2nd attempt
failed_pdfs2    = []
save_index      = 0
BATCH_SAVE_INTERVAL = 5   # or reuse the constant you had before
ATTEMPT_TAG     = "try_2" # suffix for all output filenames

def extract_text_safe(fp):
    with open(fp, 'rb') as data:
        out = BytesIO()
        extract_text_to_fp(data, out, laparams=LAParams(word_margin=0.2))
        return out.getvalue().decode()

for i, batch in enumerate(retry_batches, start=1):
    batch_start = time.time()
    print(f"[TRY-2] Batch {i}/{len(retry_batches)} – {len(batch)} DOIs")

    try:
        # ---- identical steps as first pass ----
        with open(DOI_FILE, 'w') as fh:
            fh.writelines([doi + "\n" for doi in batch])

        time.sleep(10)
        subprocess.run(
            f'python -m PyPaperBot --doi-file={DOI_FILE} --dwn-di="{PDF_DIR}" --restrict="1"',
            shell=True,
            timeout=BATCH_TIMEOUT
        )

        try:
            results = pd.read_csv(RESULTS_FILE).dropna(subset=['PDF Name'])
        except Exception as exc:
            print(f"  ✗ could not read {RESULTS_FILE}: {exc}")
            failed_batches2.append(batch)
            continue

        for doi, pdf in results.set_index('DOI')['PDF Name'].items():
            pdf_path = os.path.join(PDF_DIR, pdf)
            try:
                with concurrent.futures.ThreadPoolExecutor(max_workers=1) as ex:
                    fut = ex.submit(extract_text_safe, pdf_path)
                    text = fut.result(timeout=PDF_TIMEOUT)

                full_texts[doi] = cut_unrelevant_info(text)

            except concurrent.futures.TimeoutError:
                print(f"  ✗ timeout on {doi}")
                failed_pdfs2.append(doi)
            except Exception as exc:
                print(f"  ✗ error on {doi}: {exc}")
                failed_pdfs2.append(doi)

        subprocess.run(f'rm -f {PDF_DIR}/*.pdf', shell=True)

        # ---- periodic dump ----
        if i % BATCH_SAVE_INTERVAL == 0:
            out = f'full_texts_v2_part_{save_index}_{ATTEMPT_TAG}.json'
            with open(out, 'w') as fh:
                json.dump(full_texts, fh)
            print(f"  ✓ wrote {out} ({len(full_texts)} docs)")
            full_texts.clear()
            save_index += 1

    except subprocess.TimeoutExpired:
        print(f"  ✗ batch timed out")
        failed_batches2.append(batch)
    except Exception as exc:
        print(f"  ✗ unexpected error: {exc}")
        failed_batches2.append(batch)

# ---- flush any remainder ----
if full_texts:
    out = f'full_texts_v2_part_{save_index}_{ATTEMPT_TAG}.json'
    with open(out, 'w') as fh:
        json.dump(full_texts, fh)
    print(f"  ✓ wrote {out} ({len(full_texts)} docs)")

# ---- save the new failure logs ----
if failed_batches2:
    with open('failed_batches_try_2.json', 'w') as fh:
        json.dump(failed_batches2, fh)
    print(f"Saved {len(failed_batches2)} still-failing batches → failed_batches_try_2.json")

if failed_pdfs2:
    with open('failed_pdfs_try_2.json', 'w') as fh:
        json.dump(failed_pdfs2, fh)
    print(f"Saved {len(failed_pdfs2)} still-failing PDFs    → failed_pdfs_try_2.json")

print("Second-pass retry finished.")
```

```python
#!/usr/bin/env python3
# -----------------------------------------------------------
# 3rd-pass: split each still-failing batch into 4 sub-batches
#           and retry them individually
# -----------------------------------------------------------
import os, json, time, subprocess, concurrent.futures, pandas as pd
from pathlib import Path
from io import BytesIO
from pdfminer.high_level import extract_text_to_fp
from pdfminer.layout import LAParams

SUB_BATCH_FACTOR    = 4        # split each failing batch into 4
OUTPUT_ATTEMPT_TAG  = 'try_3'
BATCH_SAVE_INTERVAL = 5        # dump JSON after this many sub-batches

# ---------- helper -----------------------------------------------------------
def extract_text_safe(fp):
    with open(fp, 'rb') as data:
        buf = BytesIO()
        extract_text_to_fp(data, buf, laparams=LAParams(word_margin=0.2))
        return buf.getvalue().decode()

def chunk(lst, n):
    """Yield successive n-sized chunks from lst."""
    for i in range(0, len(lst), n):
        yield lst[i:i + n]

# ---------- load still-failing batches --------------------------------------
if not Path('failed_batches_try_2.json').exists():
    print("Nothing to retry – 'failed_batches_try_2.json' not found.")
    raise SystemExit

with open('failed_batches_try_2.json') as fh:
    parent_batches = json.load(fh)

print(f"[TRY-3] Loaded {len(parent_batches)} parent batches from failed_batches_try_2.json")

# ---------- retry state ------------------------------------------------------
full_texts      = {}
failed_batches3 = []   # list of *sub-batches* (list[DOI]) still failing
failed_pdfs3    = []   # list of individual DOI strings
sub_counter     = 0    # counts sub-batches overall
save_index      = 0

# ---------- main loop --------------------------------------------------------
for p_idx, parent in enumerate(parent_batches):
    # split parent batch into four roughly-equal sub-batches
    sub_size = max(1, len(parent) // SUB_BATCH_FACTOR)
    sub_batches = list(chunk(parent, sub_size))

    for s_idx, sub in enumerate(sub_batches):
        sub_counter += 1
        t0 = time.time()
        print(f"[TRY-3] Parent:{p_idx} Sub:{s_idx} ({len(sub)} DOIs)")

        try:
            # --- write DOIs -----------------------------------------------
            with open(DOI_FILE, 'w') as fh:
                fh.writelines([d + "\n" for d in sub])

            # --- small pause & run PyPaperBot -----------------------------
            time.sleep(10)  # shorter pause than v2
            subprocess.run(
                f'python -m PyPaperBot --doi-file="{DOI_FILE}" --dwn-di="{PDF_DIR}" --restrict="1"',
                shell=True,
                timeout=BATCH_TIMEOUT
            )

            # --- read results --------------------------------------------
            try:
                res = pd.read_csv(RESULTS_FILE).dropna(subset=['PDF Name'])
            except Exception as exc:
                print(f"  ✗ CSV error: {exc}")
                failed_batches3.append(sub)
                continue

            # --- per-PDF extraction --------------------------------------
            for doi, pdf in res.set_index('DOI')['PDF Name'].items():
                pdf_path = os.path.join(PDF_DIR, pdf)
                try:
                    with concurrent.futures.ThreadPoolExecutor(max_workers=1) as ex:
                        fut = ex.submit(extract_text_safe, pdf_path)
                        text = fut.result(timeout=PDF_TIMEOUT)
                    full_texts[doi] = cut_unrelevant_info(text)
                except concurrent.futures.TimeoutError:
                    print(f"  ✗ timeout on {doi}")
                    failed_pdfs3.append(doi)
                except Exception as exc:
                    print(f"  ✗ error on {doi}: {exc}")
                    failed_pdfs3.append(doi)

            # --- clean tmp PDFs ------------------------------------------
            subprocess.run(f'rm -f {PDF_DIR}/*.pdf', shell=True)

            # --- periodic save -------------------------------------------
            if sub_counter % BATCH_SAVE_INTERVAL == 0:
                fname = (f'full_texts_v3_parent{p_idx}_sub{s_idx}_'
                         f'part{save_index}_{OUTPUT_ATTEMPT_TAG}.json')
                with open(fname, 'w') as fh:
                    json.dump(full_texts, fh)
                print(f"  ✓ wrote {fname} ({len(full_texts)} docs)")
                full_texts.clear()
                save_index += 1

            print(f"  ✓ completed in {time.time() - t0:.1f}s")

        except subprocess.TimeoutExpired:
            print("  ✗ sub-batch timed out")
            failed_batches3.append(sub)
        except Exception as exc:
            print(f"  ✗ unexpected error: {exc}")
            failed_batches3.append(sub)

# ---------- flush remainder --------------------------------------------------
if full_texts:
    fname = f'full_texts_v3_parentX_subY_part{save_index}_{OUTPUT_ATTEMPT_TAG}.json'
    with open(fname, 'w') as fh:
        json.dump(full_texts, fh)
    print(f"  ✓ wrote final {fname} ({len(full_texts)} docs)")

# ---------- write new failure logs ------------------------------------------
if failed_batches3:
    with open('failed_batches_try_3.json', 'w') as fh:
        json.dump(failed_batches3, fh)
    print(f"Saved {len(failed_batches3)} sub-batches → failed_batches_try_3.json")

if failed_pdfs3:
    with open('failed_pdfs_try_3.json', 'w') as fh:
        json.dump(failed_pdfs3, fh)
    print(f"Saved {len(failed_pdfs3)} PDFs → failed_pdfs_try_3.json")

print("Third-pass retry finished.")
```
