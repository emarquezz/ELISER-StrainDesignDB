```python
def flatten_level_3_list(level_3_list):
    if len(level_3_list[0][0][0])!=1:
        level_2_list = [inner for outer in level_3_list for inner in outer]
        return level_2_list
    else :
        return level_3_list
```

```python
organism_dataframes = []
for organism_id in org_KEDD_ID:
    organism_id = str(organism_id)
    try:
        # Open the file
        kegg_info = pd.read_csv(f'../../KEGG_genes/{organism_id}.csv')
        kegg_info = kegg_info.set_index('KO')
        organism_dataframes.append(kegg_info)
        #print(kegg_info.head())
    except:
        # Get the file
        #print('getting_new')
        url  = f"https://www.genome.jp/kegg-bin/download_htext?htext={organism_id}00001&format=htext&filedir=kegg/brite/{organism_id}"
        raw_info = requests.get(url)
        text_info = StringIO(raw_info.text)

        # extract important information
        gene_info = []
        for line in text_info.readlines():
            if line.startswith('D'):
                filtered_line = line.rstrip()[7:]
                gene_info.append(filtered_line.split('\t'))

        kegg_info = pd.DataFrame(gene_info,columns=[f'Gene_{organism_id}','KO'])

        kegg_info.loc[:,f'Gene_{organism_id}'] = organism_id+':'+kegg_info.loc[:,f'Gene_{organism_id}'].str.split(' ').str[0]+';'
        kegg_info.KO = kegg_info.KO.str.split(' ').str[0]
        kegg_info.set_index('KO',inplace=True)
        kegg_info.drop_duplicates(inplace=True)
        kegg_info = kegg_info.groupby(level=0).sum()
        organism_dataframes.append(kegg_info)
        kegg_info.to_csv(f'../../KEGG_genes/{organism_id}.csv')
```

```python
full_info = pd.concat(organism_dataframes,axis=1)
```

```python
def buscar_magia(lista):
    for gene_kegg in lista:
        org = gene_kegg.split(':')[0]
        #gene = gene.split(':')[1]
        if f'Gene_{org}' in full_info:
            kegg_db = full_info.loc[:,f'Gene_{org}'].dropna().str.rstrip(';')
            kegg_db = kegg_db.str.split(';').explode()
            new_gene = kegg_db.index[kegg_db==gene_kegg]
            if not new_gene.empty:
                return(new_gene[0])
```
