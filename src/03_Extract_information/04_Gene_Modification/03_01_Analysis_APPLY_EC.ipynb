# Process Gene Lists


	1.	Remove papers without gene data  
	2.	Fix nested list structure  
	3.	Extract organism + gene + EC number combos  

```python
# Keep only rows with valid gene extraction
genes_df = full_text_df.dropna(subset=["Genes"])

genes_df.head()
```

```python
def flatten_gene_list(nested_list):
    """
    Some files contain a level 3 nest like: [[[pairs]]]
    If so, flatten to a level 2 list: [[pairs]]
    """
    if len(nested_list[0][0][0]) != 1:
        return [inner for outer in nested_list for inner in outer]
    else:
        return nested_list

# Apply
genes_df["Genes"] = genes_df["Genes"].apply(flatten_gene_list)
```

```python
# Explode into individual gene entries
expanded_genes = genes_df.explode("Genes").reset_index(drop=True)

# Keep only organism + gene pair
gene_table = expanded_genes.loc[:, ["Organism", "Genes"]]
gene_table.head()
```

```python
# Keep rows where the EC field is non-empty and non-zero
genes_with_ec = gene_table.loc[
    ~gene_table.Genes.str[-1].isin([0, "", []])
]

genes_with_ec.head()
```

```python
ec_table = genes_with_ec.copy()

# EC list
ec_table["EC_search"] = (
    ec_table.Genes.str[-1]
    .str.rstrip(";")
    .str.split(";")
    .apply(set)  # unique ECs
)

# Gene name
ec_table["Gene"] = ec_table.Genes.str[0]

ec_table = ec_table[["Organism", "Gene", "EC_search"]]
ec_table.columns = ["Organism", "Genes", "EC_search"]
ec_table.head()
```

# Process EC Table

```python
processed_ec_df = process_ec_df(ec_table)

processed_ec_df.head()
```

```python
processed_ec_df.shape
```

```python
# Expand EC lists into rows
ec_expanded = processed_ec_df.explode("EC_search").drop_duplicates()

# Normalize organism field
ec_expanded["Organism"] = ec_expanded["Organism"].str.capitalize()

# Final dedupe
ec_expanded = ec_expanded.drop_duplicates()

ec_expanded.head()
```

```python
# Reset index so Title becomes a column
modifications_df_reset = modifications_df.reset_index().rename(columns={"index": "Title"})

# Merge full text with modifications
merged_df = full_text_df.merge(modifications_df_reset, on="Title", how="left")

merged_df = merged_df[
    ["Title", "Organism", "Genes", "Gene_Source", "Modifications"]
]

merged_df.head()
```

# Map Gene Modifications to EC Modifications

```python
merged_df["EC_Modifications"] = merged_df.apply(
    lambda row: map_modifications_to_ec_fixed(
        row,
        ec_expanded,
        verbose=False,
        allow_cross_organism_fallback=False
    ),
    axis=1
)

merged_df.head()
```

```python
merged_df["EC_Modifications_Normalized"] = merged_df["EC_Modifications"].apply(
    extract_and_normalize_dict
)

merged_df.head()
```
