## <span style="color: #2171b5">Specific dates</span>

```python
# Engineering related and between specific dates 2000 and 2025
# Generate PubMed-compatible date search terms for engineering-related queries
def generate_date_batches(start_year=2000, end_year=2025):
    date_batches = []
    for year in range(start_year, end_year + 1):
        date_batches.extend([
            f"{year}/01/01:{year}/03/31[dp]",
            f"{year}/04/01:{year}/06/30[dp]",
            f"{year}/07/01:{year}/09/30[dp]",
            f"{year}/10/01:{year}/12/31[dp]",
        ])
    return date_batches

# Example usage
terms_with_dates = generate_date_batches()
```

# <span style="color: #08519c">Make multiple batches to get the info</span>

```python
import csv

full_article_ids = []

for year in range(2000, 2026):  # inclusive of 2025
    filename = f"article_ids_{year}.csv"
    try:
        with open(filename, newline="") as f:
            reader = csv.reader(f)
            next(reader)  # Skip header
            for row in reader:
                if row:  # Ensure it's not an empty row
                    full_article_ids.append(row[0])
    except FileNotFoundError:
        print(f"Warning: {filename} not found — skipping.")

# Deduplicate
full_article_ids = list(set(full_article_ids))

print(f"Total unique article IDs collected: {len(full_article_ids)}")
```

```python
full_article_ids.sort()  # Optional

with open(article_ids_file, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["ArticleID"])
    for aid in full_article_ids:
        writer.writerow([aid])
```
