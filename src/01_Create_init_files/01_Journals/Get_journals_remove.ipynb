```python
import sys
sys.path.append("../..")

import pandas as pd

from file_management import check_save_file, get_files_dir
_, INPUT_DIR, OUTPUT_DIR = get_files_dir()

from text_analysis import similar_words
```

```python

import urllib.request, urllib.parse, urllib.error
import ssl

ctx = ssl.create_default_context()
ctx.check_hostname = False
ctx.verify_mode = ssl.CERT_NONE
```

```python
journals_url = 'https://ftp.ncbi.nih.gov/pubmed/J_Medline.txt'
```

```python
key_words_avoid = pd.read_csv(INPUT_DIR+'/Journals/Avoid/key_words_avoid.txt', header=None, names = ['Words'])
key_words_avoid = key_words_avoid.Words.str.strip("'")
```

```python
# Check synonyms if you want to add more words
words = [ 'botany', 'neuro', 'viral', 
           'virus', 'virology', 'optical', 'tissue',
           'biomaterial', 'polymers', 'nano', 'disease',
           'immunity', 'cancer', 'annals', 'injuries',
           'immunology', 'clinic', 'therapy', 'orthopaedic', 
           'healthcare', 'microscopy', 'medical', 'neuron',
         'surgery', 'oncology', 'cardiology', 'material']

synonyms = 20
for word in words:
    print(synonyms, 'Similar words to',word+':\n\t',similar_words(word, synonyms),'\n\n')
```

```python
s1 = pd.Series(['odonto','allergo','nervos','anaesthes','anato','management',
                'chirurgica','derma','hepato','rheuma','skin','urolo','geronto'])
key_words_avoid = pd.concat([key_words_avoid,s1], ignore_index=True)
```

```python
key_words_avoid = pd.concat([key_words_avoid,s1], ignore_index=True)
```

```python
html = urllib.request.urlopen(journals_url, context=ctx)
neg_journals_ids = []
avoid_journal = False

for line in html:
    line = line.decode()
    line = line.lower()
    
    for word in key_words_avoid:
        if word in line:
            avoid_journal = True
    
    if avoid_journal  and 'nlmid:' in line:
        avoid_journal = False
        neg_journals_ids.append(line.split(':')[1].strip())
    elif not avoid_journal and 'journaltitle' in line:
        print(line)
```

```python
# Separate the journals in batches
batches_len = 2000
batched_art_ids = [neg_journals_ids[i * batches_len:(i + 1) * batches_len] 
                   for i in range((len(neg_journals_ids) + batches_len- 1) // batches_len )]
```

```python
# Save file
check_save_file(pd.DataFrame(batched_art_ids).T, 'list_journals_avoid.json','Journals/Avoid', input_dir = True)
```
